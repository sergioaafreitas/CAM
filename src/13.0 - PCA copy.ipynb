## Objetivo e dados

Reduzir quatro medidas a duas componentes sem usar o rótulo.

**Pergunta-guia:** Qual a proporção de variância preservada pelas duas componentes?

Execute as células na ordem e interprete os resultados antes de fazer o exercício.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
import pandas as pd

# Carregar o conjunto de dados de íris
iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names

# Aplicar PCA para reduzir de 4 para 2 componentes principais
pca = PCA(n_components=2)
X_r = pca.fit_transform(X)

# Criar um DataFrame para facilitar a visualização
df = pd.DataFrame(X_r, columns=['PC1', 'PC2'])
df['target'] = y
df['species'] = df['target'].apply(lambda i: target_names[i])

# Visualizar os dados transformados
plt.figure(figsize=(10, 7))
sns.scatterplot(x='PC1', y='PC2', hue='species', data=df, palette='Set1', s=100, alpha=0.7)
plt.title('PCA de íris: redução para 2 componentes principais')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend(loc='best', title='Espécies')
plt.grid(True)
plt.show()
```

## Investigação complementar

Compare a nova medida com a figura ou o resultado anterior. Explique qualquer diferença observada.

```python
from sklearn.preprocessing import StandardScaler
X_padronizado = StandardScaler().fit_transform(X)
pca_padronizado = PCA(n_components=2).fit(X_padronizado)
print('Variância explicada sem escala:', pca.explained_variance_ratio_.round(3))
print('Variância explicada com escala:', pca_padronizado.explained_variance_ratio_.round(3))
```

## Interpretação e limites

A projeção depende da escala das variáveis e não mede desempenho de classificação.
